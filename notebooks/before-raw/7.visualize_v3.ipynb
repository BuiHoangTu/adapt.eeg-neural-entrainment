```python
from adapt_eeg.data_reader import read_raw_data


data = read_raw_data("../data/raw/datasets_4Hz")
len(data)
```

```python
import random 

sample = random.choice(data).data
sample
```

```python
a = sample.plot(
    # duration=10,
    n_channels=5,
    scalings={"eeg": 2e-6},  # ±100 µV
    # remove_dc=False,
    # proj=False,
)
```

```python
import matplotlib.pyplot as plt

# raw is your existing mne.io.Raw object
raw_eeg = sample

spectrum = raw_eeg.compute_psd(
    method="welch",
    fmin=1,
    fmax=12,
    n_fft=512,
    n_overlap=256,
)

b = spectrum.plot(
    average=True,
    dB=True,
)

plt.show()
```
