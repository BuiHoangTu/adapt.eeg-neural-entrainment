```python
from importlib import import_module

from adapt_eeg.auto_annotation.stress_identifier import calculate_inter_stress_distances, detect_stressed_syllables
from adapt_eeg.auto_annotation.textgrid_annotation_fix.common import (
    derive_lined_audio,
)
from adapt_eeg.constants import POEMS_CONFIG

poem_id = 0

module = import_module(
    f"adapt_eeg.auto_annotation.textgrid_annotation_fix.poem{poem_id+1}"
)
textgrid_eol_indices = module.textgrid_eol_indices

poem_config = POEMS_CONFIG[poem_id]
lined = derive_lined_audio(
    poem_config["audio_url"],
    poem_config["textgrid_url"],
    textgrid_eol_indices,
    intro_end_ms=poem_config["intro_end"],
)

stresses = detect_stressed_syllables(lined)
line_dists = calculate_inter_stress_distances(lined)

pretty = [[round(x, 3) for x in row] for row in line_dists]
# pretty
```

```python
import numpy as np

avgs = []
stds = []

for i, row in enumerate(line_dists):
    avg = np.mean(row)
    std = np.std(row)
    
    avgs.append(avg)
    stds.append(std)
    
    print(f"avg={avg:.3f}, std={std:.3f}")
```

```python
from adapt_eeg.utils.audio_stress_visualization import waveform_export, browser_open


file = f"poem {poem_id+1} stresses.html"
waveform_export(lined.audio, [s.nucleus for s in stresses], file)
browser_open(file)
```

```python
from matplotlib import pyplot as plt


plt.hist(stds)
plt.show()
```
