```python
from importlib import import_module

from adapt_eeg.auto_annotation.classes import SyllableIdentifyParams
from adapt_eeg.auto_annotation.syllable_identifier import syllablize_audio
from adapt_eeg.auto_annotation.textgrid_annotation_fix.common import (
    derive_eol_times_from_audio,
)
from adapt_eeg.constants import POEMS_CONFIG

poem_id = 0

module = import_module(
    f"adapt_eeg.auto_annotation.textgrid_annotation_fix.poem{poem_id+1}"
)
textgrid_eol_indices = module.textgrid_eol_indices

poem_config = POEMS_CONFIG[poem_id]
eol_times = derive_eol_times_from_audio(
    poem_config["audio_url"],
    poem_config["textgrid_url"],
    textgrid_eol_indices,
    intro_end_ms=poem_config["intro_end"],
)

config_syl = SyllableIdentifyParams()

auto_syl = syllablize_audio(
    poem_config["audio_url"],
    config_syl, 
    line_boundaries=eol_times,
)
```
