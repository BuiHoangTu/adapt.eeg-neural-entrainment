```python
from importlib import import_module

from adapt_eeg.auto_annotation.textgrid_annotation_fix.common import (
    derive_eol_times_from_audio,
)
from adapt_eeg.constants import POEMS_CONFIG

poem_id = 1

module = import_module(f"adapt_eeg.auto_annotation.textgrid_annotation_fix.poem{poem_id+1}")
textgrid_eol_indices = module.textgrid_eol_indices

poem_config = POEMS_CONFIG[1]
eol_times = derive_eol_times_from_audio(
    poem_config["audio_url"],
    poem_config["textgrid_url"],
    textgrid_eol_indices,
    intro_end_ms=poem_config["intro_end"],
)
```

```python
from adapt_eeg.auto_annotation.syllable_identifier import _load_audio


_, audio, *_ = _load_audio(poem_config["audio_url"])
audio = audio[poem_config["intro_end"]:]
```

```python
from pathlib import Path

from adapt_eeg.utils.audio_stress_visualization import waveform_render

HTML_OUT = Path("waveform_eol.html")


html = waveform_render(audio, eol_times, uid=HTML_OUT.name)
HTML_OUT.write_text(html, encoding="utf-8")

print(f"Saved to {HTML_OUT}")
```

```python
import webbrowser

file_uri = HTML_OUT.resolve().as_uri()
webbrowser.open(file_uri, new=2)  # new=2 forces a new tab
```
